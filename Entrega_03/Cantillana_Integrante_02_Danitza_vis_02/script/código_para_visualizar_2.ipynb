```python
import pandas as pd
import altair as alt

# 1. Cargar la base de datos limpia con el separador correcto
df = pd.read_csv('/BASE_DE_DATOS_CALIDAD_ZONAS_VERDES_OK.csv', sep=';')

# 2. Convertir la columna 'CALIDAD' a número flotante limpiando comas decimales
df['CALIDAD'] = df['CALIDAD'].astype(str).str.replace(',', '.').astype(float)

# 3. Calcular el promedio de calidad por cada comuna y ordenarlo
promedio_calidad = df.groupby('COMUNA')['CALIDAD'].mean().reset_index()

# 4. Crear el gráfico de barras optimizado con Altair
grafico = alt.Chart(promedio_calidad).mark_bar(color='#2b5c8f').encode(
    x=alt.X('CALIDAD:Q', title='Promedio de Índice de Calidad'),
    y=alt.Y('COMUNA:N', sort='-x', title='Comuna (Ordenada de mayor a menor calidad)'),
    tooltip=['COMUNA', 'CALIDAD']
).properties(
    title='Promedio de Calidad de Zonas Verdes por Comuna',
    width=650,
    height=600  # Aumentamos la altura para que respiren bien los nombres de las comunas
)

# Mostrar el gráfico
grafico
```

```python
# Forzar el guardado y comprobación del archivo HTML
grafico.save('vis_01.html')

import os
if os.path.exists('vis_01.html'):
    print("¡El archivo vis_01.html está guardado y listo en tu Colab!")
else:
    print("Hubo un problema al guardarlo.")
```
